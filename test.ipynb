```python
import pandas as pd
```

```python
output_sorted = pd.read_csv('test/sorted_output.csv')

# contar o quanstas vezes cada linha diferente aparece
output_sorted['count'] = 1
output_sorted = output_sorted.groupby(['ID_a_m', 'ID_b_M', 'ID\'', 'a_m', 'b_M', 'f']).sum().reset_index()

# filtrar as linhas com count maior que 1
output_sorted = output_sorted[output_sorted['count'] > 1].reset_index(drop=True)

# sort by 'f' descending
output_sorted = output_sorted.sort_values(by='f', ascending=True).reset_index(drop=True)

output_sorted
```
